# Methods

Short and long Nanopore sequencing reads for the Alteromonas macleodii genome were provided by Dr. Joseph Orofino. Quality filtering of long reads was performed with Filtlong v0.3.0, retaining reads with a minimum length of 500 bases and the top 90% ranked by quality. Filtered reads were compressed and stored in FASTQ format. Quality assessment of short reads was performed with FastQC v0.12.1.

Long reads were assembled using Flye v2.9.6 without a reference genome. The assembly was indexed with Bowtie2 v2.5.4 and short reads were aligned to the indexed assembly using the same tool. Alignments were sorted with Samtools v1.22.1. The draft assembly was then corrected and polished with Pilon v1.24, using the assembled genome as the --genome input and aligned short reads as the --frags input. The process was allocated 32 GB of memory using the -Xmx32G flag.

Assembly statistics and genome completeness were assessed with QUAST v5.3.0 and BUSCO v6.0.0. The reference genome for Alteromonas macleodii was downloaded using the NCBI Datasets CLI v18.6.0 and compared to the initial unpolished assembly. QUAST was also used to compare the reference genome with the polished assembly from Pilon. BUSCO was run on the final polished assembly.

Finally, Prokka v1.14.6 was used to annotate the polished genome. The entire workflow was coordinated with Nextflow v25.04.6 and executed on the Massachusetts Green High Performance Computing Center, completing in approximately 1.5 hours.

# Results

Using the described pipeline, the Alteromonas macleodii genome was successfully assembled. The assembly contained 4,627,922 base pairs with 5,830,642 reads. Of these, 5,687,054 reads were mapped. Fragment orientation was consistent, with FR orientation at 100%, and the mean insert size was 389 base pairs. Pilon reported 51x coverage and confirmed 99.9% of the original bases, correcting only 8 SNPs and one deletion. Several local continuity breaks associated with tandem repeats remained unresolved.

QUAST was used to assess assembly quality relative to the reference. The polished genome had a genome fraction of 98.429%, a duplication ratio of 1.001, 19 misassemblies, 18.18 misassemblies per 100 kbp, a total length of 4,627,920 base pairs, and a GC content of 44.7%. Compared to the unpolished assembly, the only difference was a slightly higher misassembly rate of 18.33 per 100 kbp. This indicates that polishing had little effect on large scale structural accuracy, and the Flye draft assembly was already of high quality.

BUSCO (Benchmarking Universal Single Copy Orthologs) was used to evaluate assembly completeness against a set of conserved orthologs from the OrthoDB database. BUSCO classifies matches as complete (C), duplicated (D), fragmented (F), or missing (M), reporting the total number of orthologs as n.

For the polished assembly, BUSCO reported C:99.6% [S:99.5%, D:0.1%], F:0.0%, M:0.4%, n:1828. This indicates that nearly all expected orthologs were present, with only a very small fraction missing and almost no duplication. The absence of fragmented genes and the high percentage of single copy completeness support the conclusion that the assembly is both highly accurate and nearly complete.

Overall, the pipeline generated a high quality assembly. Across polishing, completeness assessment, and structural evaluation, only minimal errors were observed. Pilon confirmed 99.9% of bases, QUAST reported a genome fraction of 98.429%, and BUSCO reported 99.6% completeness. These results demonstrate that the assembly is accurate, complete, and structurally consistent.
